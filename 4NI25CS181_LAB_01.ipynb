```python
import numpy as np
import pandas as pd

print(np.__version__)
print(pd.__version__)
```

```python
marks = np.array([78,85,92,67,88])
print("marks:", marks)
print("type:",type(marks))
```

Array Operations

```python
print("Number of dimensions:",marks.ndim)
print("shape:", marks.shape)
print("Number of elements:", marks.size)
```

Numpy Operations

```python
marks = np.array([78,85,92,67,88])

print("original:", marks)
print("Add 5:", marks +5)
print("multiply by 2:", marks * 2)
print("divide by 2:", marks/2)
print("square:", marks **2)
```

```python
print("Sum:",marks.sum())
print("Mean:",marks.mean())
print("maximum:",marks.max())
print("minimum:",marks.min())
```

numpy indexing

```python
print("First mark:", marks[0])
print("Second mark:", marks[1])
print("last mark:", marks[-1])
```

numpy slicing

```python
print("index 1 to 3:", marks[1:4])
print("first three:", marks[:3])
print("from index 2:", marks[2:])
print("every second element:", marks[::2])
```

two-dimensional numpy array

```python
student_marks = np.array([
    [78,85,92],
    [67,88,76],
    [90,95,89]
])

print(student_marks)
print("shape:", student_marks.shape)
```

```python
print("mean of all values:", student_marks.mean())
print("column wise mean:", student_marks.mean(axis=0))
print("row wise mean:", student_marks.mean(axis=1))
```

pandas dataframe

a dataframe is a 2d tabular data structure in pandas

```python
data = {
    "Name": ["anu","rahul","kiran","priya"],
    "branch": ["cse","aiml","cse","aiml"],
    "marks": [85,91,76,88]
}

df = pd.DataFrame(data)
df
```

inspecting a dataframe

```python
print("shape:", df.shape)
print("\ncolumns:")
print(df.columns)
```

```python
df.head()
```

```python
df.tail()
```

```python
df.info()
```

```python
df.describe()
```

selecting columns

```python
print("marks column:")
print(df["marks"])
print("\nbranch and marks:")
print(df[["branch","marks"]])
```

```python
print("first row:")
print(df.iloc[0])

print("\nsecond row")
print(df.iloc[1])

print("\nthird row")
print(df.iloc[0:3])
```

```python
print("first 2 rows and first  2columns:")
print(df.iloc[0:2,0:2])
```

```python
high_scorers = df[df["marks"]>90]
high_scorers
```

```python
print("students from cse:")
display(df[df["branch"]=="cse"])

print("\nstudents scoring 85 or above:")
display(df[df["marks"]>= 85])
```
